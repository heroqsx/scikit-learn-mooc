# Wrap-up for tree model

Open the dataset ames_housing_no_missing.csv

```python
import pandas as pd

ames_housing = pd.read_csv(
    "../datasets/ames_housing_no_missing.csv",
    na_filter=False,  # required for pandas>2.0
)
target_name = "SalePrice"
data = ames_housing.drop(columns=target_name)
target = ames_housing[target_name]
```

To simplify this exercise, we only use the numerical features defined below:

```python
numerical_features = [
    "LotFrontage", "LotArea", "MasVnrArea", "BsmtFinSF1", "BsmtFinSF2",
    "BsmtUnfSF", "TotalBsmtSF", "1stFlrSF", "2ndFlrSF", "LowQualFinSF",
    "GrLivArea", "BedroomAbvGr", "KitchenAbvGr", "TotRmsAbvGrd", "Fireplaces",
    "GarageCars", "GarageArea", "WoodDeckSF", "OpenPorchSF", "EnclosedPorch",
    "3SsnPorch", "ScreenPorch", "PoolArea", "MiscVal",
]

data_numerical = data.loc[:, numerical_features]
```

```python
data_numerical.head()
```

tune of linear model with 10 fold cross validation

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import cross_validate
```

```python
# create pipeline with standard scaler + linear regression
model = make_pipeline(StandardScaler(), LinearRegression())
cv_results = cross_validate(model, data_numerical, target, cv=10, return_estimator=True)
cv_results["test_score"]
```

```python
# create pipeline with standard scaler + tree regression
model = make_pipeline(StandardScaler(), DecisionTreeRegressor(random_state=1))
cv_results = cross_validate(model, data_numerical, target, cv=10, return_estimator=True)
cv_results["test_score"]
```

```python


pipeline = make_pipeline(StandardScaler(), DecisionTreeRegressor(random_state=2))
param_grid = {'decisiontreeregressor__max_depth': range(1, 16)}
grid_search = GridSearchCV(pipeline, param_grid, cv=10)
cv_results = cross_validate(grid_search, data_numerical, target, cv=10, return_estimator=True)
cv_results["test_score"]
```

```python
for i in range(len(cv_results["estimator"])):
    print(cv_results["estimator"][i].best_params_)
```
